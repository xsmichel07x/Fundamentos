```python
from IPython.display import display, HTML

# Lista de alumnos activos en el sistema
alumnos = ["26040200", "26040201", "26040202", "26040203", "26040204"]

# Cada valor en la lista de computadoras tiene su lista correspondiente
computadoras = [
    ["PC-LC1-01", "LC1", "Python, VS Code", "Disponible"],
    ["PC-LC1-02", "LC1", "Python, Java, VS Code", "Prestada"],
    ["PC-LC1-03", "LC1", "AutoCAD, MATLAB", "Reservada"],
    ["PC-LC2-01", "LC2", "Python, VS Code", "Disponible"],
    ["PC-LC2-02", "LC2", "Python, Java, VS Code", "Prestada"],
    ["PC-LC2-03", "LC2", "AutoCAD, MATLAB", "Disponible"],
    ["PC-LC3-01", "LC3", "Python, VS Code", "Disponible"],
    ["PC-LC3-02", "LC3", "Python, Java, VS Code", "Prestada"],
    ["PC-LC3-03", "LC3", "AutoCAD, MATLAB", "Reservada"],
    ["PC-LC3-04", "LC3", "Python, VS Code", "Mantenimiento"]
]

prestamos = []

def render_box(titulo, contenido_html, color_borde="#ffb6c1", color_fondo="#fff0f5"):
    display(HTML(f'''
    <div style="background-color: {color_fondo}; border: 2px solid {color_borde}; border-radius: 12px; padding: 15px; margin: 10px 0; max-width: 650px; font-family: 'Segoe UI', sans-serif; color: #4a3b40; box-shadow: 0 2px 8px rgba(255,182,193,0.3);">
        <div style="font-weight: bold; color: #d81b60; border-bottom: 1.5px dashed {color_borde}; padding-bottom: 6px; margin-bottom: 10px; font-size: 15px;">
            {titulo}
        </div>
        <div style="font-size: 13.5px; line-height: 1.6;">
            {contenido_html}
        </div>
    </div>
    '''))

def consultar_computadoras():
    html = "<table style='width: 100%; border-collapse: collapse; font-size: 13px;'>"
    for i in range(len(computadoras)):
        html += f"""
        <tr style='border-bottom: 1px solid #f8bbd0;'>
            <td style='padding: 4px;'><b>Aula:</b> {computadoras[i][1]}</td>
            <td style='padding: 4px;'><b>ID:</b> {computadoras[i][0]}</td>
            <td style='padding: 4px;'><b>Software:</b> {computadoras[i][2]}</td>
            <td style='padding: 4px;'><b>Estado:</b> <span style='color: #ad1457; font-weight: 600;'>{computadoras[i][3]}</span></td>
        </tr>
        """
    html += "</table>"
    render_box("--- COMPUTADORAS ---", html)

def buscar_sw():
    programa = input("\nIngrese el software que desea buscar: ")
    encontrado = False
    html = "<table style='width: 100%; border-collapse: collapse; font-size: 13px;'>"
    for i in range(len(computadoras)):
        if programa.lower() in computadoras[i][2].lower():
            html += f"""
            <tr style='border-bottom: 1px solid #f8bbd0;'>
                <td style='padding: 4px;'><b>ID:</b> {computadoras[i][0]}</td>
                <td style='padding: 4px;'><b>Aula:</b> {computadoras[i][1]}</td>
                <td style='padding: 4px;'><b>Software:</b> {computadoras[i][2]}</td>
                <td style='padding: 4px;'><b>Estado:</b> {computadoras[i][3]}</td>
            </tr>
            """
            encontrado = True
    html += "</table>"

    if encontrado:
        render_box("--- RESULTADOS ---", html)
    else:
        render_box("--- RESULTADOS ---", "<span style='color: #880e4f;'>No se encontraron computadoras con ese software.</span>", "#f8bbd0", "#fdf2f8")

def solicitar_prestamo():
    render_box("===SOLICITAR PRESTAMO===", "<span style='color: #880e4f;'>Ingrese los datos en la consola a continuación:</span>")
    num_control = input("Numero de control del alumno: ")
    if num_control in alumnos:
        programa = input("Que programa necesita: ")
        for i in range(len(computadoras)):
            if programa.lower() in computadoras[i][2].lower():
                if computadoras[i][3] == "Disponible":
                    computadoras[i][3] = "Prestada"
                    msg = f"Se asignó la computadora <b>{computadoras[i][0]}</b> que esta en el aula <b>{computadoras[i][1]}</b>"
                    render_box("CONFIRMACIÓN", msg, "#c8e6c9", "#f1f8e9")
                    prestamo = [num_control, programa, computadoras[i][1], computadoras[i][0]]
                    prestamos.append(prestamo)
                    break
                else:
                    render_box("AVISO", f"Si se encontro el sofware pero esta <b>{computadoras[i][3]}</b>", "#ffe0b2", "#fff8e1")
    else:
        render_box("ERROR", "EL alumno no existe en la lista", "#ffcdd2", "#ffebee")

def devolver_computadora():
    render_box("===DEVOLVER COMPUTADORA====", "<span style='color: #880e4f;'>Ingrese los datos en la consola a continuación:</span>")
    id_pc = input("Dame el ID de la computadora: ")
    existe = False
    for i in range(len(computadoras)):
        if computadoras[i][0] == id_pc:
            existe = True
            if computadoras[i][3] == "Prestada" or computadoras[i][3] == "Reservada":
                computadoras[i][3] = "Disponible"
                render_box("CONFIRMACIÓN", f"La computadora <b>{id_pc}</b> ya esta disponible", "#c8e6c9", "#f1f8e9")
                break
            else:
                render_box("AVISO", f"La computadora <b>{id_pc}</b> no esta prestada/reservada", "#ffe0b2", "#fff8e1")
    if not existe:
        render_box("ERROR", f"No existe la computadora <b>{id_pc}</b>", "#ffcdd2", "#ffebee")

def agregar_computadora():
    render_box("--- AGREGAR COMPUTADORA ---", "<span style='color: #880e4f;'>Ingrese los datos en la consola a continuación:</span>")
    id_computadora = input("ID de la computadora: ")
    aula = input("Aula: ")
    programas = input("Software instalado: ")
    nueva_computadora = [id_computadora, aula, programas, "Disponible"]
    computadoras.append(nueva_computadora)
    render_box("ÉXITO", "Computadora agregada correctamente.", "#c8e6c9", "#f1f8e9")

def mostrar_prestamos():
    if len(prestamos) == 0:
        render_box("--- HISTORIAL DE PRESTAMOS ---", "No hay prestamos registrados.")
    else:
        html = "<table style='width: 100%; border-collapse: collapse; font-size: 13px;'>"
        for i in range(len(prestamos)):
            html += f"""
            <tr style='border-bottom: 1px solid #e1bee7;'>
                <td style='padding: 4px;'><b>Alumno:</b> {prestamos[i][0]}</td>
                <td style='padding: 4px;'><b>Software:</b> {prestamos[i][1]}</td>
                <td style='padding: 4px;'><b>Aula:</b> {prestamos[i][2]}</td>
                <td style='padding: 4px;'><b>ID:</b> {prestamos[i][3]}</td>
            </tr>
            """
        html += "</table>"
        render_box("--- HISTORIAL DE PRESTAMOS ---", html, "#e1bee7", "#f3e5f5")

# MENÚ PRINCIPAL
opcion = 0
while opcion <= 7:
    display(HTML('''
    <div style="background-color: #fff0f5; border: 2.5px solid #ffb6c1; border-radius: 15px; padding: 18px; max-width: 420px; font-family: 'Segoe UI', sans-serif; color: #4a3b40; box-shadow: 0 4px 10px rgba(255, 182, 193, 0.3);">
        <div style="text-align: center; border-bottom: 1.5px dashed #f8bbd0; padding-bottom: 8px; margin-bottom: 12px;">
            <h2 style="color: #d81b60; font-size: 18px; margin: 0; font-weight: bold; letter-spacing: 1px;">SISTEMA DE PRÉSTAMOS</h2>
        </div>
        <div style="font-size: 14px; line-height: 1.8; color: #4a3b40;">
            <b>1.</b> Consultar computadoras<br>
            <b>2.</b> Buscar por software<br>
            <b>3.</b> Solicitar préstamo<br>
            <b>4.</b> Devolver computadora<br>
            <b>5.</b> Crear reservación<br>
            <b>6.</b> Agregar computadora<br>
            <b>7.</b> Historial de prestamos<br>
            <b>8.</b> Salir
        </div>
    </div>
    '''))

    try:
        opcion = int(input("\nSeleccione una opción: "))
    except ValueError:
        render_box("ERROR", "Opción no válida.", "#ffcdd2", "#ffebee")
        continue

    if opcion == 1:
        consultar_computadoras()
    elif opcion == 2:
        buscar_sw()
    elif opcion == 3:
        solicitar_prestamo()
    elif opcion == 4:
        devolver_computadora()
    elif opcion == 5:
        render_box("--- CREAR RESERVACIÓN ---", "Función en desarrollo.", "#f8bbd0", "#fff0f5")
    elif opcion == 6:
        agregar_computadora()
    elif opcion == 7:
        mostrar_prestamos()
    elif opcion == 8:
        render_box("SISTEMA", "Gracias por utilizar el sistema.", "#f8bbd0", "#fff0f5")
    else:
        render_box("ERROR", "Opción no válida.", "#ffcdd2", "#ffebee")
```
