```python
tabla = int(input("Que tabla de multiplicar quieres ver? "))
limiteinicial = int(input("Desde que numero quieres empezar? "))
limitefinal = int(input("Hasta que numero quieres llegar? "))
while limitefinal < limiteinicial:
  print("El limite final no puede ser menor que el inicial")
  limitefinal = int(input("Hasta que numero quieres llegar? "))
incrementos= int(input("De cuanto en cuanto quieres que vaya? "))
 for i in range(limiteinicial, limitefinal + 1, incrementos):
    print(tabla, "x", i, "=", tabla * i)
```

# **Tema de listas**

```python
alumnos = ["Ana", "Luis", "Carlos", "Juan"]
print(alumnos)
print(alumnos[0])
print(alumnos[3])
print(alumnos[5]) #esta instruccion marcara error pq el indice o posicion no existe
```

```python
#Modificar un valor de la lista en un indice especifico
alumnos[1]= "Maria Luisa"
print(alumnos)
```

```python
#El metodo append agrega un elemento al final de la lista
alumnos.append("Karla")
print(alumnos)
```

```python
#El metodo insert agrega un elemento a la lista en una posicion especifica
alumnos.insert(2, "Cristina")
print(alumnos)
```

```python
#remove elimina un elemento especificando el valor del elemento, debe coincidir exactamente
alumnos.remove("Ana")
alumnos.remove("ana") #este marcaria error pq no coincide exactamente
```

```python
print(alumnos)
```

```python
#pop elimina el elemento o item en la posicion especificada
alumnos.pop(2)
print(alumnos)
```

```python
#del tambien elimina el elemento o item en la posicion especificada
del alumnos[3]
print(alumnos)
```

```python
#len permite conocer cuantos elementos tiene una lista
print("Elementos en la lista de alumnos:", len(alumnos))
```

```python
for x in alumnos:
  print(x)
```

```python
print(alumnos)
for x in alumnos:
  print(x)
  if x == "Ana":
    break #break forza a salir del ciclo, siempre y cuando se cumpla la condicion del if
```
