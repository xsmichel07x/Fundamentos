```python
# Declaramos una funcion llamada saludar
# que no hace nada, hasta que la mandamos llamar
def saludar():
  print("Hola")
```

```python
# Definir un parametro llamado nombre y puede recibir un
# valor que usaremos dentro de la funcion
def saludar2(nombre):
  print("Hola", nombre, "como estas?")
```

```python
saludar2("Sofia")
```

```python
def tablas_de_multiplicar(x):
  for i in range(10):
    print(x, "por", i, "=", x*i)
```

```python
tablas_de_multiplicar(3)
tablas_de_multiplicar(5)
```
