```python
# ============================================
# SISTEMA DE PRÉSTAMOS
# Versión 1 - Programación basica y uso de listas
# ============================================


# Lista de alumnos activos en el sistema
alumnos = [ "26040200", "26040201", "26040202", "26040203", "26040204"]


# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras = ["PC-LC1-01", "PC-LC1-02", "PC-LC1-03",
                "PC-LC2-01", "PC-LC2-02", "PC-LC2-03",
                "PC-LC3-01", "PC-LC3-02", "PC-LC3-03", "PC-LC3-04"]

aulas = ["LC1", "LC1", "LC1",
         "LC2", "LC2", "LC2",
         "LC3", "LC3", "LC3", "LC3"]

software = ["Python, VS Code", "Python, Java, VS Code", "AutoCAD, MATLAB",
            "Python, VS Code", "Python, Java, VS Code", "AutoCAD, MATLAB",
            "Python, VS Code", "Python, Java, VS Code", "AutoCAD, MATLAB", "Python, VS Code"]

estados = [ "Disponible", "Prestada", "Reservada",
            "Disponible", "Prestada", "Reservada",
            "Disponible", "Prestada", "Reservada", "Mantenimiento" ]

# MENÚ PRINCIPAL
# Usamos un ciclo While que se repetira indefinidamente hasta que el usuario
# teclee la opcion 7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion <= 6:
    print("\n===============================")
    print("       SISTEMA DE PRÉSTAMOS")
    print("===============================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Salir")
    print("===============================")
    opcion = int(input("Seleccione una opción: "))

# 1. Consultar computadoras
# si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
# toda la lista de computadoras e imprime los valores de las demás listas
if opcion == 1:
    print("\n--- COMPUTADORAS ---")
    for i in range(len(computadoras)):
        print("Aula:", aulas[i])
        print("ID:", computadoras[i])
        print("Software:", software[i])
        print("Estado:", estados[i])
        print("-----------------------")

# 2. Buscar por software
elif opcion == 2:
    programa = input( "\nIngrese el software que desea buscar: " )
    print("\n--- RESULTADOS ---")
    encontrado = False
    for i in range(len(computadoras)):
        # la funcion (lower) convierte a minuscula el texto, de esta forma
        # nos aseguramos que el texto a buscar y la informacion de las listas
        # este en minuscula y no tengamos problema de que no encunentra un programa
        # porque este escrito de forma diferente usando mayusculas y minusculas
        if programa.lower() in software[i].lower():
            print("ID:", computadoras[i])
            print("Aula:", aulas[i])
            print("Software:", software[i])
            print("Estado:", estados[i])
            encontrado = True

    if encontrado == False:
        print("No se encontraron computadoras con ese software.")

# 3. Solicitar préstamo
elif opcion == 3:
    print("\n--- SOLICITAR PRÉSTAMO ---")
    num_control = input( "\nIngrese el número de control: " )
    encontrado1 = False
    # Verificar si existe el alumno en la lista o BD
    if num_control in alumnos:
        # Si existe, continuamos con el procedimiento
        encontrado1 = True
        id_computadora = input("Ingrese el ID de la computadora: " )
        encontrado2 = False
        for i in range(len(computadoras)):
            if computadoras[i] == id_computadora:
                encontrado2 = True
                if estados[i] == "Disponible":
                    estados[i] = "Prestada"
                    print("Préstamo registrado correctamente.")
                    print("Computadora:", computadoras[i])
                    print("Aula:", aulas[i])
                else:
                    print("La computadora no está disponible.")
        if encontrado2 == False:
            print("La computadora no existe.")
    if encontrado1 == False:
        print("El alumno no existe.")
```
