```python
# ========================================
# SISTEMA DE PRÉSTAMOS Y RESERVACIÓN
# ========================================

# Lista de alumnos activos en el sistema
alumnos = [ { "num_control" : "26040200", "nombre" : "Superman"       , "carrera" : "Ingenieria en Inteligencia Artificial"  } ,
            { "num_control" : "26040201", "nombre" : "Batman"         , "carrera" : "Ingenieria en Sistemas Computacionales" } ,
            { "num_control" : "26040202", "nombre" : "Spiderman"      , "carrera" : "Ingenieria en Sistemas Computacionales" } ,
            { "num_control" : "26040203", "nombre" : "Thor"           , "carrera" : "Ingenieria en Inteligencia Artificial"  } ,
            { "num_control" : "26040203", "nombre" : "Ironman"        , "carrera" : "Ingenieria en Sistemas Computacionales" } ,
            { "num_control" : "26040204", "nombre" : "Capitan America", "carrera" : "Ingenieria en Inteligencia Artificial"  } , ]

# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras = [ { "id" : "PC-LC1-01", "aula" : "LC1", "programas" : "Python, VS Code"      , "estado" : "Disponible"    } ,
                 { "id" : "PC-LC1-02", "aula" : "LC1", "programas" : "Python, Java, VS Code", "estado" : "Prestada"      } ,
                 { "id" : "PC-LC1-03", "aula" : "LC1", "programas" : "AutoCAD, MATLAB"      , "estado" : "Reservada"     } ,
                 { "id" : "PC-LC2-01", "aula" : "LC2", "programas" : "Python, VS Code"      , "estado" : "Disponible"    } ,
                 { "id" : "PC-LC2-02", "aula" : "LC2", "programas" : "Python, Java, VS Code", "estado" : "Prestada"      } ,
                 { "id" : "PC-LC2-03", "aula" : "LC2", "programas" : "AutoCAD, MATLAB"      , "estado" : "Disponible"    } ,
                 { "id" : "PC-LC3-01", "aula" : "LC3", "programas" : "Python, VS Code"      , "estado" : "Disponible"    } ,
                 { "id" : "PC-LC3-02", "aula" : "LC3", "programas" : "Python, Java, VS Code", "estado" : "Prestada"      } ,
                 { "id" : "PC-LC3-03", "aula" : "LC3", "programas" : "AutoCAD, MATLAB"      , "estado" : "Reservada"     } ,
                 { "id" : "PC-LC3-04", "aula" : "LC3", "programas" : "Python, VS Code"      , "estado" : "Mantenimiento" } , ]

prestamos = [ { "num_control" : "26040201", "programa" : "python", "aula" : "LC1", "id" : "PC-LC1-01" } ,
              { "num_control" : "26040202", "programa" : "java"  , "aula" : "LC1", "id" : "PC-LC1-02" } ,
              { "num_control" : "26040201", "programa" : "matlab", "aula" : "LC2", "id" : "PC-LC2-01" } ,
              { "num_control" : "26040203", "programa" : "python", "aula" : "LC3", "id" : "PC-LC3-01" } ,
              { "num_control" : "26040204", "programa" : "python", "aula" : "LC3", "id" : "PC-LC3-02" } , ]

necesidades = [ { "programa" : "Photoshop", "fecha" : "12-102026 07:30:00" }, ]

def consultar_computadoras():
    print("\n   .｡.:*･°  COMPUTADORAS  °･*:.｡.   ")
    for i in range(len(computadoras)):
        print("Aula:", computadoras[i]["aula"],"| ID:", computadoras[i]["id"] ,
              "| Software:", computadoras[i]["programas"]," |Estado:", computadoras[i]["estado"] )

def buscar_sw():
  programa = input( "\nIngrese el software que desea buscar: ")
  print("\n   .｡.:*･°  RESULTADOS  °･*:.｡.   ")
  encontrado = False
  for i in range(len(computadoras)):
      if programa.lower() in computadoras[i]["programas"].lower():
          print("ID:", computadoras[i]["id"], "| Aula:", computadoras[i]["aula"],
                "| Software:", computadoras[i]["programas"], "| Estado:", computadoras[i]["estado"])
          encontrado = True
  if encontrado == False:
      print("No se encontraron computadoras con ese software.")

from datetime import datetime
def solicitar_prestamo():
    print("\n   ꒰ 🎀 SOLICITAR PRESTAMO 🎀 ꒱   ")
    num_control = input("Numero de control del alumno: ")
    if num_control in alumnos[0]:
        programa = input("Que programa necesita: ")
        encontrado = False
        for i in range(len(computadoras)):
            if programa.lower() in computadoras[i]["programas"].lower():
                if computadoras[i]["estado"] == "Disponible":
                    computadoras[i]["estado"] = "Prestada"
                    print("Se asignó la computadora",
                        computadoras[i]["id"],
                        "que esta en el aula",
                        computadoras[i]["aula"])
                    prestamo = [num_control, programa, computadoras[i]["aula"], computadoras[i]["id"]]
                    prestamos.append(prestamo)
                    encontrado = True
                    break
                else:
                    print("Si se encontro el sofware pero esta", computadoras[i]["estado"])
        if not encontrado:
            print("\nEl programa no está disponible.")
            respuesta = input("¿Desea agregar este programa a la lista de necesidades? (si/no): ")
            if respuesta.lower() in ["si", "sí"]:
                fecha_hora_actual = datetime.now().strftime("%d-%m-%Y %H:%M:%S")
                necesidades.append([programa, fecha_hora_actual])
                print(f"El programa '{programa}' ha sido agregado a la lista de necesidades ({fecha_hora_actual}).")
    else:
        print("EL alumno no existe en la lista")

def devolver_computadora():
  print("\n   ꒰ 🎀 DEVOLVER COMPUTADORA 🎀 ꒱   ")
  id_pc = input("Dame el ID de la computadora: ")
  existe = False
  for i in range( len(computadoras) ):
    if computadoras[i]["id"] == id_pc:
      existe = True
      if computadoras[i]["estado"]== "Prestada" or computadoras[i]["estado"]=="Reservada":
        computadoras[i]["estado"] = "Disponible"
        print("La computadora", id_pc, "ya esta disponible")
        break
      else:
        print("La computadora", id_pc, "no esta prestada/reservada")
  if existe == False:
    print("No existe la computadora", id_pc)

def agregar_computadora():
  print("\n   .｡.:*･°  AGREGAR COMPUTADORA  °･*:.｡.   ")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_computadora = [id_computadora, aula, programas, "Disponible"]
  computadoras.append(nueva_computadora)
  print("\nComputadora agregada correctamente.")

def mostrar_prestamos():
  print("\n   .｡.:*･°  HISTORIAL DE PRESTAMOS  °･*:.｡.   ")
  if len(prestamos) == 0:
    print("No hay prestamos registrados.")
  for i in range(len(prestamos)):
    print("Alumno:", prestamos[i][0], "| Software:", prestamos[i][1],
          "| Aula:", prestamos[i][2], "| ID:", prestamos[i][3])

def mostrar_necesidades():
  print("\n   .｡.:*･°  HISTORIAL DE NECESIDADES  °･*:.｡.   ")
  if len(necesidades) == 0:
    print("No hay necesidades registradas.")
  else:
    for i in range(len(necesidades)):
      print("Programa:", necesidades[i][0], "| Fecha y hora:", necesidades[i][1])

def mostrar_estadisticas():
  print("\n   .｡.:*･°  ESTADISTICAS DE PRESTAMOS  °･*:.｡.   ")
  if len(prestamos) == 0:
    print("No hay préstamos registrados para generar estadísticas todavía.")
    return
  conteo_programas = {}
  for p in prestamos:
      prog = p[1]
      conteo_programas[prog] = conteo_programas.get(prog, 0) + 1
  print("\n   ✿ Préstamos por Programa ✿   ")
  for prog, cant in conteo_programas.items():
      print(f"- {prog}: {cant} vez/veces")
  conteo_carreras = {}
  lista_alumnos = alumnos[0]
  for p in prestamos:
      n_control = p[0]
      carrera_alumno = "Carrera No Registrada"
      for i in range(0, len(lista_alumnos), 3):
          if lista_alumnos[i] == n_control:
              carrera_alumno = lista_alumnos[i+2]
              break
      conteo_carreras[carrera_alumno] = conteo_carreras.get(carrera_alumno, 0) + 1
  print("\n   ✿ Préstamos por Carrera ✿   ")
  for carrera, cant in conteo_carreras.items():
      print(f"- {carrera}: {cant} vez/veces")

# MENÚ PRINCIPAL
opcion = 0
while opcion <= 10:
    print("\n   ୨୧ ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~ ୨୧")
    print("        ♡ SISTEMA DE PRÉSTAMOS ♡")
    print("   ୨୧ ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~ ୨୧")
    print(" 1. Consultar computadoras")
    print(" 2. Buscar por software")
    print(" 3. Solicitar préstamo")
    print(" 4. Devolver computadora")
    print(" 5. Crear reservación")
    print(" 6. Agregar computadora")
    print(" 7. Historial de prestamos")
    print(" 8. Historial de necesidades")
    print(" 9. Estadisticas")
    print(" 10. Salir")
    print("   ୨୧ ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~ ୨୧")
    opcion = int(input("Seleccione una opción: "))

    if opcion == 1:
      consultar_computadoras()

    elif opcion == 2:
      buscar_sw()

    elif opcion == 3:
      solicitar_prestamo()

    elif opcion == 4:
      devolver_computadora()

    elif opcion == 5:
      print("\n   .｡.:*･°  CREAR RESERVACIÓN  °･*:.｡.   ")

    elif opcion == 6:
      agregar_computadora()

    elif opcion == 7:
      mostrar_prestamos()

    elif opcion == 8:
      mostrar_necesidades()

    elif opcion == 9:
      mostrar_estadisticas()

    elif opcion == 10:
        print("\nGracias por utilizar el sistema.")
        break
    else:
        print("\nOpción no válida.")
```
